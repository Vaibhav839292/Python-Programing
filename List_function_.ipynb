```python
#List Function



a = ["Thor","Hulk","Ironman","Caption America"]

print(a)

#to  find the lenght of list

print(len(a))

#to count an a=occurance of particular element

print(a.count("Hulk"))

#to add to the list
a.append("Spiderman")
print(a)

#to add toa specific location

a.insert(1,"Vison")
print(a)

#to ermove frome a lst

a.remove("Hulk")
print(a)

#to remove from a certain location
print(a.pop(1))
print(a)


```

```python
a = ["Thor","Hulk","Ironman","Caption America"]

#to create a copy of a list

b = a.copy()
print(b)

b = a.copy()
print(b)

#to access an element

print(a.index("Ironman"))





#to entend an element

c =("Spiderman","Vision")
a.extend(c)
print(a)

#to reverse the list

a.reverse()
print(a)

#to sort the list

a.sort()
print(a)

#to clear all the data from the list

a.clear()
print(a)



```

```python
#List Comprehension

l1 = [30,40,50,60]
l2 =[]

for i in l1:
  if i>45:
   l2.append(i)

print(l1,"\n",l2)

l3 = [i for i in l1]
print(l3)
```
