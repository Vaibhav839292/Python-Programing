```python
a = {"ironman","Hulk","Thor","Caption America"}
print (a)
print(type(a))
for x in a :
  print(x)
```

```python
#Fuction of sets

a = {"Ironman","Hulk","Thor","Caption America"}

#add
a.add("spiderman")
print(a)

#Pop
a.pop()
print(a)

#remove
a.remove("Thor")
print(a)

#Discard
a.discard("hulk")
print(a)

#Copy
b = a.copy()
print(b)



```

```python
a = {"Ironman","Hulk","Thor","Caption America"}
b = {"Superman","Batman","Wonder-Woman"}
c = {"Hulk","Thor"}

#isdisjoint
print(a.isdisjoint(b))

#issubset
print(c.issubset(a))

#issuperset
print(b.issuperset(a))

#update
a.update(b)
print(a)

#union
print(a.union(b))

#clear
a.clear()
print(a)

#Difference
print(a.difference(c))
#diffrence update
print(a.difference_update(c))
print(a)

#intersection
print(a.intersection(c))
#intersection update
a.intersection_update(c)
print(a)

#symetric diffrence
print(a.symmetric_difference(c))
print(a)

a.symmetric_difference_update(c)
print(a)

```
