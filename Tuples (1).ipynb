```python
#Problem Solving


#Convert the following dictionary into JSON format.

import json

student_data = {"name":"David","age":13,"marks":87}
print(type(student_data))
data = json.dumps(student_data)
print(data)
print(type(data))


#Access the value of age from the given data.


# The 'data' variable already holds the JSON string. json.loads expects a string, not a dict.
loaded_data = json.loads(data)
# Access dictionary elements using square brackets and call print as a function.
print(loaded_data["age"])

#pretty print following JSON data.
student_data = {"name":"David","age":13,"marks":87}
data = json.dumps(student_data,indent = 4, separators =(",","="))
print(data)

#sort the following JSON keys and write them into a file.
f = open("demo.json","w")
json.dump(student_data,f, indent = 4, sort_keys = True)
print("the data has been added to the file ")




```

```python
student = {"name":"Jhon","class":"6th","roll_no":23}
#get
x = student.get("name")
print(x)

#ITEM

# Corrected: use items() instead of item() and print its result
print(student.items())

#Keys
b= student.keys()
print(b)

#values
c = student.values()
print(c)

#Copy
d = student.copy()
print(d)

#setdefault
x = student.setdefault("roll_no",24)
print(x)
```
